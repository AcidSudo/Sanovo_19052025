```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv('/Users/tyco/Desktop/MADS/sanovo_19052025/data/processed/cleaned_data.csv')
df.head(5)
```

```python
df_agg = df.groupby('MerDateDay').apply(
    lambda x: pd.Series({
        'Total settings changed': ((x['MerType'] == 15)).sum(),
        'Total stops': ((x['MerType'] == 8) & (x['MerNumber'] == 0)).sum()
    })
)

df_agg.head(3)
```

```python
df['MerDate'] = pd.to_datetime(df['MerDate'])
df_agg_1h = df.set_index('MerDate').resample('1h').apply(
    lambda x: pd.Series({
        'Total settings changed': ((x['MerType'] == 15)).sum(),
        'Total stops scheduled': ((x['MerType'] == 8) & (x['MerNumber'] == 0)& (x['MerSubNumber'] == 1)).sum(),
        'Total stops conveyor full': ((x['MerType'] == 8) & (x['MerNumber'] == 0)& (x['MerSubNumber'] == 11)).sum(),
        'Total stops lane pressure': ((x['MerType'] == 8) & (x['MerNumber'] == 0)& (x['MerSubNumber'] == 2)).sum(),

        'Total stops': ((x['MerType'] == 8) & (x['MerNumber'] == 0)).sum()
    })
).reset_index()

df_agg_1h.head()
```
